```python
import pandas as pd
import matplotlib.pyplot as plt

def plot_magnetic_levitation(csv_path, xname, yname, xlabel, ylabel, 
                                reference_value=None, output='plot.pgf'):
    df = pd.read_csv(csv_path)
    
    plt.figure(figsize=(3,2.5))
    plt.plot(df[xname], df[yname])
    
    if reference_value is not None:
        plt.axhline(y=reference_value, color='r', linestyle='--')
    
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True, alpha=0.3)
    
    plt.savefig(output, bbox_inches='tight')
    plt.close()
```

```python
def plot_magnetic_levitation_mc(csv_path, xname, xlabel, ylabel, 
                                reference_value=None, output='plot.pgf', scale=1.0):
    df = pd.read_csv(csv_path)
    
    plt.figure(figsize=(3, 2.5))
    
    # Identifica as colunas de simulação
    sim_cols = [col for col in df.columns if col != xname]
    
    # 1. Plota as trajetórias individuais multiplicando pelo 'scale' (ex: 1000 para mm)
    for col in sim_cols:
        plt.plot(df[xname], df[col] * scale, color='gray', alpha=0.4, linewidth=0.5)
    
    # 2. Plota a média multiplicada pelo scale
    mean_val = df[sim_cols].mean(axis=1) * scale
    plt.plot(df[xname], mean_val, color='blue', linewidth=0.8, label='Média')
    
    # 3. Referência (o reference_value já deve ser passado na unidade final, ex: 100)
    if reference_value is not None:
        plt.axhline(y=reference_value, color='r', linestyle='--', 
                    linewidth=1.0, label='Ref.')
    
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True, alpha=0.1, linestyle=':')
    plt.tight_layout()
    
    plt.savefig(output, bbox_inches='tight')
    plt.close()
```

```python
plot_magnetic_levitation_mc("mc_corrente_smci_p.csv", "Tempo", "Tempo ($s$)", "Corrente ($A$)", output="corrente_ismc_p.pgf")
```

```python
plot_magnetic_levitation_mc("mc_tensao_smci_p.csv", "Tempo", "Tempo ($s$)", "Tensão ($V$)", output="tensao_ismc_p.pgf")
```

```python
plot_magnetic_levitation_mc("mc_velocidade_smci_p.csv", "Tempo", "Tempo ($s$)", "Velocidade ($m/s$)", output="velocidade_ismc_p.pgf")
```
